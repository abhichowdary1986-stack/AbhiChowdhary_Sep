# Exception handling

- Two blocks

- One is try block and another one is except block

- original code will write in try block

- if any error is there in original code it will catch in except block

- **try and except** - concept is used to handle the errors without crashing the code written

- 

** Finally Block **


```python
try:
    print('I have maje a good tea')
except Exception as e:
    print(e)

finally:
    print('Okay! clean the mess')
```
